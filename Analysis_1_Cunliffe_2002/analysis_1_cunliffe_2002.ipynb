## Analysis 1: Topical Clindamycin vs. Combination Therapy (Cunliffe et al., 2002)

```python
from pathlib import Path

# Save generated files in organized folders
DATA_DIR = Path("data")
FIGURES_DIR = Path("figures")
DATA_DIR.mkdir(exist_ok=True)
FIGURES_DIR.mkdir(exist_ok=True)

import pandas as pd

# Data extracted from Paper #3, Table III
data = {
    "Week": [4, 8, 12, 16],
    "Combination Gel": [-0.5, -0.2, 0.1, -0.2],
    "Clindamycin Alone": [-0.5, -0.2, 0.8, 1.2]
}

df = pd.DataFrame(data)

display(df)
```

```python
# Paper #3, Table III data visualize
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.plot(df["Week"], df["Combination Gel"],
         marker="o", linewidth=2, label="Combination Gel")

plt.plot(df["Week"], df["Clindamycin Alone"],
         marker="s", linewidth=2, label="Clindamycin Alone")

plt.axhline(0, color="gray", linestyle="--")

plt.title("Changes in Clindamycin-Resistant C. acnes Counts")
plt.xlabel("Treatment Week")
plt.ylabel("Mean log10 Change from Baseline")

plt.xticks([4, 8, 12, 16])
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(FIGURES_DIR / "antibiotic_resistance.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# Paper #3, Table V: Clinical efficacy
clinical = pd.DataFrame({
    "Week": [4, 8, 12, 16],
    "Combination Gel": [-27.1, -40.7, -59.1, -52.7],
    "Clindamycin Alone": [-7.9, -14.0, -33.1, -27.5]
})

plt.figure(figsize=(8, 5))

plt.plot(clinical["Week"], clinical["Combination Gel"],
         marker="o", linewidth=2, label="Combination Gel")

plt.plot(clinical["Week"], clinical["Clindamycin Alone"],
         marker="s", linewidth=2, label="Clindamycin Alone")

plt.axhline(0, color="gray", linestyle="--")

plt.title("Changes in Total Acne Lesions During Treatment")
plt.xlabel("Treatment Week")
plt.ylabel("Median Percent Change from Baseline (%)")

plt.xticks([4, 8, 12, 16])
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(FIGURES_DIR / "clinical_efficacy.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# Calculate the difference in acne lesion reduction
clinical["Difference (percentage points)"] = (
    clinical["Clindamycin Alone"] - clinical["Combination Gel"]
)

display(clinical)

# Week 16 comparison
week16 = clinical[clinical["Week"] == 16].iloc[0]

print("Week 16 results:")
print(f"Combination Gel: {week16['Combination Gel']}%")
print(f"Clindamycin Alone: {week16['Clindamycin Alone']}%")
print(f"Difference: {week16['Difference (percentage points)']:.1f} percentage points")
```

```python
df.to_csv(DATA_DIR / "resistance_data.csv", index=False)
clinical.to_csv(DATA_DIR / "clinical_efficacy_data.csv", index=False)

print("Both CSV files saved successfully!")
```

```python
# Calculate differences between treatment groups
resistance_comparison = df.copy()

resistance_comparison["Group Difference"] = (
    resistance_comparison["Clindamycin Alone"]
    - resistance_comparison["Combination Gel"]
)

display(resistance_comparison)

# Visualize differences over time
plt.figure(figsize=(8, 5))

plt.bar(
    resistance_comparison["Week"].astype(str),
    resistance_comparison["Group Difference"]
)

plt.title("Difference in Resistant C. acnes Changes Between Treatments")
plt.xlabel("Treatment Week")
plt.ylabel("Difference in Mean log10 Change")

plt.axhline(0, color="gray", linestyle="--")
plt.tight_layout()

plt.savefig(FIGURES_DIR / "resistance_group_difference.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# Save the additional analysis results
resistance_comparison.to_csv(
    DATA_DIR / "resistance_group_comparison.csv", index=False
)

print("Analysis saved successfully!")
```
