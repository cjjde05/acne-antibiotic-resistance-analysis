```python
import pandas as pd
import matplotlib.pyplot as plt

# Ross et al. (1997): resistant clinical isolates
df = pd.DataFrame({
    "Mutation": ["A2058G", "A2059G", "G2057A"],
    "Isolates": [13, 7, 6]
})

display(df)

plt.figure(figsize=(7, 4))
plt.bar(df["Mutation"], df["Isolates"])
plt.xlabel("23S rRNA Mutation")
plt.ylabel("Number of Resistant Isolates")
plt.title("23S rRNA Mutations in Resistant Clinical Isolates")

for i, count in enumerate(df["Isolates"]):
    plt.text(i, count + 0.2, str(count), ha="center")

plt.ylim(0, 15)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Ross et al. (1997), Table 2: erythromycin MIC (µg/mL)
# Values are intervals or limits, NOT exact point measurements.
labels = ["Wild type", "A2058G", "G2057A", "A2059G"]
ypos = [3, 2, 1, 0]

fig, ax = plt.subplots(figsize=(9, 4.5))

# ≤0.06: leftward arrow (upper bound)
ax.annotate("", xy=(0.015, 3), xytext=(0.06, 3),
            arrowprops=dict(arrowstyle="->", lw=2.5))
ax.text(0.07, 3.13, "≤0.06", fontsize=10)

# ≥512: rightward arrow (lower bound)
ax.annotate("", xy=(1000, 2), xytext=(512, 2),
            arrowprops=dict(arrowstyle="->", lw=2.5))
ax.text(512, 2.13, "≥512", fontsize=10)

# 1–2: reported MIC interval
ax.plot([1, 2], [1, 1], linewidth=5, solid_capstyle="round")
ax.plot([1, 2], [1, 1], "o")
ax.text(1.15, 1.13, "1–2", fontsize=10)

# ≥256: rightward arrow (lower bound)
ax.annotate("", xy=(1000, 0), xytext=(256, 0),
            arrowprops=dict(arrowstyle="->", lw=2.5))
ax.text(256, 0.13, "≥256", fontsize=10)

ax.set_xscale("log")
ax.set_xlim(0.01, 1500)
ax.set_ylim(-0.6, 3.65)
ax.set_yticks(ypos, labels)
ax.set_xlabel("Erythromycin MIC (µg/mL, log scale)")
ax.set_title("Reported Erythromycin MIC Bounds by 23S rRNA Mutation")
ax.grid(axis="x", which="major", alpha=0.25)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Ross et al. (1997), Table 1
species_df = pd.DataFrame({
    "Class": ["I", "II", "III", "IV", "Miscellaneous"],
    "P. acnes": [11, 0, 6, 5, 4],
    "P. granulosum": [1, 2, 0, 1, 1],
    "P. avidum": [1, 0, 0, 1, 0]
})

display(species_df)

species_df.set_index("Class").plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Resistance Phenotypes by Propionibacterium Species")
plt.xlabel("Resistance Phenotype")
plt.ylabel("Number of Isolates")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Ross et al. (1997), Table 2
# MIC ranges are shown as labels, not exact measurements.
mic_df = pd.DataFrame({
    "Mutation": ["Wild type", "A2058G", "G2057A", "A2059G"],
    "Erythromycin": ["≤0.06", "≥512", "1–2", "≥256"],
    "Josamycin": ["≤0.06", "1–16", "≤0.25", "≥256"]
})

display(mic_df)

fig, ax = plt.subplots(figsize=(9, 4))

ax.axis("off")
table = ax.table(
    cellText=mic_df.values,
    colLabels=mic_df.columns,
    loc="center",
    cellLoc="center"
)

table.auto_set_font_size(False)
table.set_fontsize(11)
table.scale(1, 2)

plt.title("Antibiotic MIC by 23S rRNA Mutation (µg/mL)")
plt.tight_layout()
plt.show()
```

### Results

Ross et al. (1997) identified three 23S rRNA mutations associated with antibiotic resistance in cutaneous propionibacteria. A2058G was found in 13 resistant clinical isolates, A2059G in 7, and G2057A in 6.

A2058G and A2059G were associated with high erythromycin MICs, while G2057A was associated with lower resistance levels. These findings suggest that different mutations in 23S rRNA are associated with different antibiotic resistance patterns.

**Limitations:** The study included multiple Propionibacterium species. MIC values were reported as ranges or bounds, so the graph does not represent exact MIC measurements.
